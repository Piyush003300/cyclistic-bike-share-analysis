```python
# Importing Libraries
import numpy as np
import pandas as pd
import glob
```

```python
# Loading Files

files = glob.glob("data/*.csv")
```

```python
# Checking if all the columns name in each table are same

ref_col = list(pd.read_csv(files[0], nrows=0).columns)
for f in files:
    curr_col = list(pd.read_csv(f,nrows=0).columns)
    print(f,"->",curr_col==ref_col)
    
```

```python
# Concat all the tables

df = pd.concat((pd.read_csv(f) for f in files), ignore_index=True)
print(df.shape)
```

```python
df.describe()
```

```python
df.sample(5)
```

```python
df.info()
```

```python
# Converting datatype to datetime

df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])
```

```python
df.info()
```

```python
df['member_casual'].value_counts()
```

```python
# New column : Ride length minutes

df['ride_length_min'] = (df['ended_at'] - df['started_at']).dt.total_seconds()/60
```

```python
df['ride_length_min'].describe()
```

```python
print("minimum ride in minutes : ", df['ride_length_min'].min(), "\nmaximum ride in hours : ",
df['ride_length_min'].max()/60)
```

```python
# Removing ride that whose length time is less then 0 and greater then 24 hrs

df_cleaned = df[df['ride_length_min'] > 0]
```

```python
df_cleaned = df_cleaned[df_cleaned['ride_length_min'] <= 1440]
```

```python
print("minimum ride in minutes : ", df_cleaned['ride_length_min'].min(), "\nmaximum ride in hours : ",
df_cleaned['ride_length_min'].max()/60)
```

```python
# Creating a column that shows day name

df_cleaned['day_of_week']= df_cleaned['started_at'].dt.day_name()
```

```python
# summary by day 


day_summary = df_cleaned.groupby(['member_casual', 'day_of_week']).agg(
    ride_count=('ride_id', 'count'),
    avg_ride_length=('ride_length_min', 'mean')
).reset_index()

print(day_summary)
```

```python
day_summary.to_csv('data/summary_by_day.csv', index=False)
```

```python
# summary by month


month_summary = df_cleaned.groupby(['member_casual', 'month']).agg(
    ride_count=('ride_id', 'count'),
    avg_ride_length=('ride_length_min', 'mean')
).reset_index()

month_summary.to_csv('data/summary_by_month.csv', index=False)
```

```python
# summary by bike type


bike_summary = df_cleaned.groupby(['member_casual', 'rideable_type']).agg(
    ride_count=('ride_id', 'count')
).reset_index()

bike_summary.to_csv('data/summary_by_bike_type.csv', index=False)
```

```python
df_cleaned = pd.read_csv('data/cleaned_trips.csv', parse_dates=['started_at', 'ended_at'])
```

```python
df_cleaned['hour'] = df_cleaned['started_at'].dt.hour
```

```python
# summary by hour


hour_summary = df_cleaned.groupby(['member_casual', 'hour']).agg(
    ride_count=('ride_id', 'count')
).reset_index()

hour_summary['rank_within_group'] = hour_summary.groupby('member_casual')['ride_count'].rank(ascending=False, method='min')

hour_summary.to_csv('data/summary_by_hour.csv', index=False)
```

```python
# summary by avg length


avg_length_summary = df_cleaned.groupby('member_casual').agg(
    avg_ride_length=('ride_length_min', 'mean')
).reset_index()

avg_length_summary.to_csv('data/summary_avg_length.csv', index=False)
```
